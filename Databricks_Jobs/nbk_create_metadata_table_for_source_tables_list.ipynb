## Drop the Metadata Table for Storing Active Source and Corresponding Ingestion Table Names

```python
sql_query = f"""
DROP TABLE IF EXISTS {catalog_name}.{schema_name}.source_tables_list_ctl
"""
spark.sql(sql_query)
```

## Create the Metadata Table for Storing Active Source and Corresponding Ingestion Table Names

```python
sql_query = f"""
CREATE TABLE IF NOT EXISTS {catalog_name}.{schema_name}.source_tables_list_ctl
(
    source_table_name STRING,
    ingestion_table_name STRING,
    active_flag STRING
)
"""
spark.sql(sql_query)
```
