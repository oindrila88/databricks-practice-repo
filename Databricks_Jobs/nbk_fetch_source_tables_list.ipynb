## Fetch Names of the Source Tables and Corresponding Ingestion Tables

```python
sql_query = f"""
SELECT              source_table_name,
                    ingestion_table_name
FROM                {catalog_name}.{schema_name}.source_tables_list_ctl
WHERE               active_flag = 'Y'
"""
df_tables_list = spark.sql(sql_query)

# First, Convert DataFrame to Python List Using "collect ()"
tables_list = df_tables_list.collect()
print(f"In List, The Source and the Ingestion Tables are : {tables_list}")

# Then, Convert Python List to Python Dictionary Using "asDict ()"
dict_tables_list = [row.asDict() for row in tables_list]
print(f"In Dictionary, The Source and the Ingestion Tables are : {dict_tables_list}")
```

## Set Up the Output for the Task that Calls this Notebook

```python
dbutils.jobs.taskValues.set(
    key = "list_of_tables",
    value = dict_tables_list
)
```
