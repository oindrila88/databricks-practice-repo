## Get Source Table Name and Ingestion Table Name as Inputs Using Text Widgets

```python
dbutils.widgets.text("source_table_name", "", "Table Name to Ingest")
source_table_name = dbutils.widgets.get("source_table_name")
print(f"Table Name to Ingest : {source_table_name}")

dbutils.widgets.text("ingestion_table_name", "", "Table Name to Create")
ingestion_table_name = dbutils.widgets.get("ingestion_table_name")
print(f"Table Name to Create : {ingestion_table_name}")
```

## Ingest Data from Source Table in Sample Schema to Ingestion Table to Desired Schema

```python
sql_query = f"""
CREATE OR REPLACE TABLE {catalog_name}.{schema_name}.{ingestion_table_name}
AS
SELECT * FROM {sample_catalog_name}.{sample_schema_name}.{source_table_name}
"""
spark.sql(sql_query)
```
